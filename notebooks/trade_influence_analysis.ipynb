# Trade Influence Indices (I, E, CWI, CWE)

Research question 2: vulnerability to foreign influence through trade.

**Indices** (partners: Australia, China, United States)
- **I** — Import index: bilateral imports / total imports
- **E** — Export index: bilateral exports / total exports
- **CWI / CWE** — Commodity-weighted import/export indices (Comtrade only; SITC-2)

IMF DOTS supplies I and E only (no commodity detail). Plots are time series by country and by partner, plus Comtrade vs IMF overlays for I and E.

This notebook runs the same pipeline as `scripts/run_trade_influence_analysis.py`.


```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import Image, display

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "trade_influence").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from trade_influence.constants import OUTPUT_CSV_DIR, OUTPUT_DIR, OUTPUT_PLOTS_DIR
from trade_influence.pipeline import run_analysis

plt.style.use("seaborn-v0_8-whitegrid")
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.4f}")

print(f"CSV dir: {OUTPUT_CSV_DIR}")
print(f"Plots dir: {OUTPUT_PLOTS_DIR}")
```
